So according the paper what is "not influence" is controlled in this dataset by who posted it.
The control group is based on known good user ids that are known not to be part of an influence campaign.

A quick look at data/raw/meysam2020_original/Tweet_IDs/TweetIDs_RandomUsers.txt reveals the tweets in this dataset consist of tweet ids, not the tweet text iself.

Per https://stackoverflow.com/questions/897107/can-i-fetch-the-tweet-from-twitter-if-i-know-the-tweets-id web scraping may be in order
Believe it or not, this still works:
twitter.com/anyuser/status/<tweetid>
for example, using one of the tweet ids from 
[twitter.com/anyuser/status/421303671155355648](https://twitter.com/anyuser/status/421303671155355648)
redirects to
https://x.com/dariocalmese/status/421303671155355648

However an X account with API privledges would be the best way to retreive these tweets, but that is hard to get today and could be very expensive.
