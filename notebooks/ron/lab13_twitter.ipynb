## Creating an X (Twitter) Account and Configuring for Python

This lab sets up an X (Twitter) account using an email address, and stores credentials in a `.env` file.

### Steps to Create an X (Twitter) Account

1. **Visit X (Twitter):**
   - Go to [twitter.com](https://twitter.com) in your web browser.

2. **Sign Up:**
   - Click on "Sign up" and enter your name, email, and date of birth.
   - Verify your email using the code sent to you.
   - Create a password when prompted.

3. **Complete Setup:**
   - Choose a username and skip additional setup steps as desired.



## Storing Credentials in a `.env` File

1. **Create a `.env` File:**
   - In your project directory, create a file named `.env`.

2. **Add Your Credentials:**
   - Open the `.env` file and add the following lines:
     ```plaintext
     XUSER=your_email@example.com
     XPASSWORD=your_password
     ```

3. **Load Credentials in Python:**
   - Run the following code to verify your credential exist in the `.env` file.
   - Verify XUSER and XPASSWORD print out.
   

````python
from soxm.Paths import Paths
from dotenv import dotenv_values

path = Paths.project()/".env"
print(path)
print("exists" if path.exists() else "does not exist")

config = dotenv_values(path)
for k, v in config.items():
    print(f'{k} key exists')
````

Verify `XUSER` and `XPASSWORD` is displayed above.

The following code will load it into python:

```{python}
from soxm.Paths import Paths
from dotenv import dotenv_values
config = dotenv_values(Paths.project()/".env")
xuser = config['XUSER']

```
